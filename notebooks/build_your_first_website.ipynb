## Step 4 - Start a browser preview

Run the next cell **once** to start a simple web server for the `website` folder. Then:
1. Open the **Ports** panel in Codespaces.
2. Find port **8000**.
3. Open the forwarded port in the browser.
4. Refresh the page each time Copilot updates your files.


````python
from pathlib import Path
import os
import socket
import subprocess
import time

pid_file = Path("website/.server.pid")
if pid_file.exists():
    old_pid = int(pid_file.read_text().strip())
    try:
        os.kill(old_pid, 0)
        print(f"Preview server is already running on port 8000 (PID {old_pid}).")
    except OSError:
        pid_file.unlink()
        old_pid = None
else:
    old_pid = None

if old_pid is None:
    process = subprocess.Popen(
        ["python3", "-m", "http.server", "8000", "--directory", "website"],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL,
        start_new_session=True,
    )
    pid_file.write_text(str(process.pid), encoding="utf-8")
    print(f"Preview server started on port 8000 (PID {process.pid}).")

for _ in range(20):
    try:
        with socket.create_connection(("127.0.0.1", 8000), timeout=1):
            print("Preview server is ready.")
            break
    except OSError:
        time.sleep(0.25)
else:
    print("The server started, but port 8000 is not ready yet. Wait a moment and open the forwarded port again.")

print("Open the forwarded port in Codespaces to preview your website.")
````

## Step 5 - Check that HTML, CSS, and JavaScript are all connected

Run the next cell after Copilot has updated your files. It performs a few beginner-friendly checks.


````python
from pathlib import Path

html = Path("website/index.html").read_text(encoding="utf-8")
css = Path("website/styles.css").read_text(encoding="utf-8")
js = Path("website/script.js").read_text(encoding="utf-8")

checks = {
    "index.html has content": bool(html.strip()),
    "styles.css is linked": "styles.css" in html,
    "script.js is linked": "script.js" in html,
    "styles.css has content": bool(css.strip()),
    "script.js has content": bool(js.strip()),
}

for label, passed in checks.items():
    print(("PASS" if passed else "NEEDS WORK"), "-", label)
````
